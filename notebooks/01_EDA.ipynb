```python
Q1 = df['Income'].quantile(0.25)
Q3 = df['Income'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[
    (df['Income'] >= lower_bound) &
    (df['Income'] <= upper_bound)
]
```

```python
df.shape
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['Income'])

plt.title("Income Boxplot After Outlier Removal")

plt.show()
```

```python
categorical_cols = ['Education', 'Marital_Status']
```

```python
df = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)
```

```python
features = [
    'Income',
    'Age',
    'Total_Spending',
    'Total_Purchases',
    'Family_Size',
    'Recency'
]

X = df[features]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
```

```python
for i in range(1, 11):

    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)
```

```python
df['Cluster'] = clusters
```

```python
df['Cluster'].value_counts()
```

```python
cluster_summary = df.groupby('Cluster')[
    [
        'Income',
        'Age',
        'Total_Spending',
        'Total_Purchases',
        'Family_Size',
        'Recency'
    ]
].mean()

cluster_summary
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
pca_df = pd.DataFrame(
    X_pca,
    columns=['PCA1', 'PCA2']
)

pca_df['Cluster'] = clusters
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='PCA1',
    y='PCA2',
    hue='Cluster',
    data=pca_df,
    palette='Set1'
)

plt.title("Customer Segments")

plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, clusters)

print("Silhouette Score:", score)
```

```python
cluster_names = {
    0: 'Budget Customers',
    1: 'Premium Customers',
    2: 'Family Customers',
    3: 'Loyal Customers'
}

df['Customer_Segment'] = df['Cluster'].map(cluster_names)
```

```python
df[['Cluster', 'Customer_Segment']].head()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Customer_Segment',
    data=df
)

plt.title("Customer Segment Distribution")

plt.xticks(rotation=15)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x='Customer_Segment',
    y='Total_Spending',
    data=df
)

plt.title("Average Spending by Segment")

plt.xticks(rotation=15)

plt.show()
```
